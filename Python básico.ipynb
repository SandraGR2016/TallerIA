
### 1. Organizar el horario de una conferencia
Estás organizando una conferencia y necesitas programar una serie de sesiones en una sala de reuniones. Dada una lista de duraciones de sesiones y un tiempo total disponible en horas, ¿cuántas sesiones caben dentro del tiempo disponible? Algunas sesiones pueden necesitar ser divididas en partes mas pequeáas.

- **Entrada**:
  - Una lista de duraciones de sesiones en minutos, por ejemplo `[30, 45, 60, 90, 15]`.
  - El tiempo total disponible en horas, por ejemplo `5 horas`.

- **Salida**:
  - Una lista de sesiones que pueden ser programadas dentro del tiempo disponible.


```python
def programar_sesiones(duraciones, horas_disponibles):
    tiempo_disponible = horas_disponibles * 60  # pasar horas a minutos
    sesiones_programadas = []

    for sesion in duraciones:
        if tiempo_disponible > 0:
            if sesion <= tiempo_disponible:
                sesiones_programadas.append(sesion)
                tiempo_disponible -= sesion
            else:
                # añadir solo la parte que cabe
                sesiones_programadas.append(tiempo_disponible)
                tiempo_disponible = 0

    return sesiones_programadas


# Ejemplo
duraciones = [30, 45, 60, 90, 15]
horas_disponibles = 3

resultado = programar_sesiones(duraciones, horas_disponibles)
print("Sesiones programadas:", resultado)
```

```python
def programar_sesiones_numpy(duraciones, horas_disponibles):
    duraciones = np.array(duraciones)
    tiempo_total = horas_disponibles * 60  # pasar a minutos

    # Suma acumulada de las sesiones
    acumulado = np.cumsum(duraciones)

    # Índices de sesiones que caben completas
    sesiones_completas = duraciones[acumulado <= tiempo_total]

    resultado = list(sesiones_completas)

    # Ver si queda tiempo para una sesión parcial
    if len(sesiones_completas) < len(duraciones):
        tiempo_usado = np.sum(sesiones_completas)
        tiempo_restante = tiempo_total - tiempo_usado
        
        if tiempo_restante > 0:
            resultado.append(tiempo_restante)

    return resultado


# Ejemplo
duraciones = [30, 45, 60, 90, 15]
horas_disponibles = 3

resultado = programar_sesiones_numpy(duraciones, horas_disponibles)
print("Sesiones programadas:", resultado)
```


### 7. Monitoreo de datos de salud
Se te han dado los datos de pasos diarios para un usuario. Tu tarea es:
- Calcular el número promedio de pasos por día.
- Encontrar el día con el mayor y menor número de pasos.
- Identificar los días en los que el usuario alcanza el objetivo de 10,000 pasos.

- **Entrada**:
  - Una lista de pasos diarios, por ejemplo `[8000, 12000, 9000, 15000, 11000, 10000, 8000, 7500, 10000]`.

- **Salida**:
  - El número promedio de pasos por día.
  - El día con el mayor y menor número de pasos.
  - La lista de días cuando el usuario alcanza el objetivo de 10,000 pasos.

```python
pasos = np.array([8000, 12000, 9000, 15000, 11000, 10000, 8000, 7500, 10000])

# Promedio
promedio = np.mean(pasos)

# Máximo y mínimo
max_pasos = np.max(pasos)
min_pasos = np.min(pasos)

# Día del máximo y mínimo (+1 porque los días empiezan en 1)
dia_max = np.argmax(pasos) + 1
dia_min = np.argmin(pasos) + 1 #arg min() Indice

# Días con al menos 10,000 pasos
objetivo = 10000
dias_objetivo = np.where(pasos >= objetivo)[0] + 1

# Resultados
print("Promedio de pasos:", promedio)
print("Día con más pasos:", dia_max, f"({max_pasos} pasos)")
print("Día con menos pasos:", dia_min, f"({min_pasos} pasos)")
print("Días que alcanzan el objetivo:", dias_objetivo)
```
