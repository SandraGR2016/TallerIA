# API-REST

En los casos en los que no existe una API tan elaborada como en Twitter, se utiliza una API basada en la arquitectura software de transferencia de estado representacional (REpresentational State Transfer)

Este protocolo define un método para recibir y enviar datos en XML o JSON bajo el protocolo HTTP. Utilizan funciones POST, GET, PUT y DELETE.

La mayoría de aplicaciones como Facebook, Youtube... cuentan con su propia API-REST.
