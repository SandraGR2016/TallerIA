## 2. Introducción a Pandas

### ¿Qué es Pandas?

**Pandas** es una librería de Python de **código abierto** diseñada para la manipulación y análisis de datos.  
Su nombre proviene de *Panel Data*, un término estadístico que hace referencia a datasets multidimensionales.  

Se ha convertido en un **estándar de facto** en la industria de la ciencia de datos e inteligencia artificial, ya que facilita el trabajo con **datos tabulares**.

---

### ¿Por qué Pandas es tan utilizado?

Antes de Pandas, trabajar con datos en Python era mucho más tedioso:
- Se usaban listas y diccionarios, que resultaban poco eficientes para operaciones complejas.
- Para cálculos numéricos existía **NumPy**, muy rápido, pero con estructuras poco expresivas para datos tabulares.

Pandas combina lo mejor de ambos mundos:
- **Velocidad** y eficiencia gracias a que está construido sobre **NumPy** y escrito en parte en **C**.
- **Expresividad** gracias a sus estructuras `Series` y `DataFrame`, que permiten representar datos de manera clara y manipulable.

En la práctica, Pandas ofrece:
- **Lectura/escritura** de datos en múltiples formatos (CSV, Excel, JSON, SQL, Parquet, HDF5...).
- **Selección y filtrado** de datos con sintaxis muy flexible.
- **Transformaciones**: aplicar funciones, renombrar columnas, crear nuevas variables.
- **Manejo de valores faltantes** de forma sencilla (`NaN`, `None`).
- **Agrupaciones y agregaciones** para análisis estadístico.
- **Operaciones vectorizadas** (mucho más rápidas que los bucles tradicionales).
- **Integración perfecta** con librerías de IA y Machine Learning (Scikit-learn, PyTorch, TensorFlow).

En resumen: **Pandas es el puente entre los datos en bruto y los modelos de IA.**

---

### ¿Qué tiene Pandas por debajo?

Para entender por qué Pandas es tan potente, conviene conocer qué hay bajo el capó:

- **NumPy**: Pandas se apoya en NumPy para el almacenamiento y cálculo eficiente de arrays multidimensionales.
- **Cython** y **C**: partes críticas de Pandas están optimizadas en bajo nivel para máxima velocidad.
- **Indexación**: cada `Series` y `DataFrame` tiene un índice que permite acceso rápido y flexible a los datos.
- **Operaciones vectorizadas**: en lugar de recorrer fila a fila, Pandas aplica operaciones a columnas enteras de forma optimizada.
- **Tipado eficiente**: Pandas maneja enteros, flotantes, cadenas, fechas, categorías... y permite ahorrar memoria con conversiones.

Esto significa que, aunque la sintaxis sea muy "amigable", Pandas está altamente optimizado para manejar **millones de filas** en memoria.

---

### ¿Por qué Pandas es clave en IA?

En un flujo de trabajo de Inteligencia Artificial, Pandas aparece en varias fases:

1. **Recolección de datos**  
   Lectura desde ficheros CSV, Excel, JSON o bases de datos.

2. **Limpieza de datos**  
   Manejo de valores nulos, duplicados, outliers.

3. **Transformación de datos**  
   Normalización, creación de nuevas variables, selección de columnas relevantes.

4. **Análisis exploratorio (EDA)**  
   Estadísticas rápidas, visualización previa con librerías como `plotly`, `matplotlib` o `seaborn`.

5. **Preparación del dataset**  
   División en train/test, exportación a NumPy arrays o tensores de PyTorch/TensorFlow.

En la práctica, **casi todos los proyectos de IA en Python comienzan con Pandas**.

## 4. Tipos de datos y conversión

### <u>Los `dtypes` en Pandas</u>

Cada columna de un **DataFrame** tiene un tipo de datos asociado (`dtype`).  
Estos tipos definen cómo se almacenan los valores en memoria y qué operaciones son posibles.

Ejemplo con el dataset Titanic reducido:

````python
df_csv = pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"))
df_csv.dtypes
````

Observaciones:
- `int64`: números enteros.
- `float64`: números decimales.
- `object`: tipo genérico en Pandas (normalmente cadenas de texto).
- `category`: tipo especial para valores repetitivos (lo veremos más abajo).
- `datetime64`: fechas y tiempos.

### <u>Conversión de tipos (casting)</u>

Podemos cambiar el tipo de una columna con `.astype()`:

````python
# Convertir survived a booleano
df_csv["survived_bool"] = df_csv["survived"].astype(bool)

# Convertir embarque a string explícito
df_csv["embarked_str"] = df_csv["embarked"].astype(str)
````

<u>Importante:</u>
> `int64` no acepta valores nulos.  
> `Int64` (con mayúscula) es un tipo Pandas que sí admite NaN.

### <u>Conversión automática al leer datos</u>

Cuando usamos `read_csv` o similares, Pandas intenta adivinar el tipo de cada columna.  
Esto funciona en muchos casos, pero a veces conviene especificar `dtype` en la lectura:

````python
df = pd.read_csv(
    os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"),
    dtype={"survived": "int8", "pclass": "category"}
)
df.dtypes
````

>- `int8`: ocupa menos memoria que `int64` (útil si los valores son pequeños).  
>- `category`: convierte directamente a tipo categórico.

### <u>El tipo `category`</u>

Las columnas con pocos valores repetidos se pueden convertir a categorías.  
Esto reduce el uso de memoria y acelera operaciones de agrupación.

Ejemplo:

````python
df["sex"] = df["sex"].astype("category")
df["embarked"] = df["embarked"].astype("category")

print(df.dtypes)
print(df["sex"].cat.categories)
````

>- Internamente, Pandas guarda las categorías como números enteros, asociados a un diccionario de etiquetas.
>- Ventaja: ahorro de memoria y rapidez en filtros y `groupby`.

### <u>Fechas y tiempos (datetime64)</u>

Otra conversión habitual es pasar texto a fechas:

````python
fechas = pd.Series(["2024-01-01", "2024-02-15", "2024-03-10"])
fechas = pd.to_datetime(fechas)
print(fechas)
````

Esto permite usar atributos y operaciones:

````python
print(fechas.dt.month)   # Mes
print(fechas.dt.weekday) # Día de la semana
````

### <u>Optimización de memoria</u>

En datasets grandes, optimizar `dtypes` puede reducir el uso de memoria drásticamente:

````python
print("Antes:")
print(df.dtypes)
print(df.memory_usage(deep=True))
print("\n"+"-"*30)

# Ejemplo de optimización
df_opt = df.copy()
df_opt["pclass"] = df_opt["pclass"].astype("int8")
df_opt["survived"] = df_opt["survived"].astype("int8")
df_opt["sex"] = df_opt["sex"].astype("category")
df_opt["embarked"] = df_opt["embarked"].astype("category")

print("\nDespués:")
print(df_opt.dtypes)
print(df_opt.memory_usage(deep=True))
````

>**Claves para optimizar:**
>- Usar int8, int16 o float32 en lugar de int64/float64 si los valores lo permiten.
>- Convertir columnas con pocos valores distintos a category.
>- Usar datetime64 para fechas.

#### ¿Cuándo usar `int8`, `int16` o `float32` en lugar de `int64`/`float64`?

Por defecto, Pandas asigna a los números enteros el tipo **`int64`** y a los decimales **`float64`**.  
Estos tipos ocupan **8 bytes por valor** y permiten representar números muy grandes, mucho más de lo que normalmente necesitamos en un dataset típico.

🔹 Tipos enteros (`int`)

- `int8` → ocupa 1 byte → valores entre **-128 y 127**  
- `int16` → ocupa 2 bytes → valores entre **-32,768 y 32,767**  
- `int32` → ocupa 4 bytes → valores entre **-2,147,483,648 y 2,147,483,647**  
- `int64` → ocupa 8 bytes → valores muy grandes (≈9e18)

👉 Recomendación:
- Si una columna contiene identificadores pequeños (ejemplo: `pclass` en Titanic, que solo puede ser 1, 2 o 3) → usar `int8`.
- Si una columna contiene recuentos modestos (ejemplo: número de hijos, `parch`, o de hermanos, `sibsp`, que nunca pasan de 10) → usar `int8` o `int16`.
- Solo usar `int64` si realmente hay números muy grandes (ejemplo: IDs enormes o valores acumulativos).

---

🔹 Tipos decimales (`float`)

- `float32` → ocupa 4 bytes → precisión de unos **7 decimales**  
- `float64` → ocupa 8 bytes → precisión de unos **15 decimales**

👉 Recomendación:
- Para columnas de dinero, precios o medidas donde no necesitamos tanta precisión → usar `float32`.
- Para cálculos financieros o científicos que requieran alta precisión → mantener `float64`.
- En Machine Learning, muchas librerías (TensorFlow, PyTorch) trabajan con `float32` por defecto, ya que ahorra memoria y es más rápido en GPU.

## 5. Limpieza y manejo de missing values

En datasets reales, es muy común encontrarse con valores **faltantes** (missing values).  
En Pandas, los valores faltantes suelen representarse como **`NaN`** (*Not a Number*) o `None`.

Detectarlos y tratarlos es una de las tareas más importantes antes de entrenar un modelo de IA.

---

### <u>Detección de Missing Values</u>

- **`isna()`** → devuelve un booleano indicando si cada valor es nulo.  
- **`notna()`** → lo contrario.

````python
df = pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"))

# Detectar valores nulos en cada celda
print(df.isna().head(10))

print("-"*100)

# Contar valores nulos por columna
print(df.isna().sum())
````

### <u>Eliminación de Missing Values</u>

- **`dropna()`** → Elimina filas o columnas con valores nulos.

````python
# Eliminar filas con al menos un valor nulo
df_sin_nulos = df.dropna()

# Eliminar columnas con valores nulos
df_sin_nulos_cols = df.dropna(axis=1)
````

````python
df.head(20)
# df_sin_nulos.head(20)
# df_sin_nulos_cols.head(20)
````

> **IMPORTANTE**:
>- Usar esta estrategia solo cuando los nulos son pocos o irrelevantes.
>- Si se eliminan demasiados, se puede perder información valiosa.

### <u>Relleno de Missing Values</u>

- **`fillna()`** → Permite reemplazar los nulos con un valor concreto.

````python
# Rellenar edad con la media
df["age_fill_mean"] = df["age"].fillna(df["age"].mean())

# Rellenar edad con la mediana
df["age_fill_median"] = df["age"].fillna(df["age"].median())

# Rellenar con un valor fijo
df["age_fill_0"] = df["age"].fillna(0)

df
````

### <u>Estrategia de imputación de Missing Values</u>

Dependiendo del tipo de dato y del problema, se pueden aplicar diferentes estrategias:

- **Eliminar filas con valores nulos** → recomendable solo si representan menos del **5%** del dataset.  

- **Reemplazar por estadísticos**:
  - **Media** → para datos numéricos distribuidos de forma normal.  
  - **Mediana** → para datos numéricos con **outliers**.  
  - **Moda** → para datos categóricos (el valor más frecuente).  

- **Forward Fill (`ffill`) / Backward Fill (`bfill`)** → arrastrar el valor anterior o posterior en la columna.  

- **Imputación avanzada** → usar algoritmos de Machine Learning para predecir los valores faltantes (ejemplo: `KNNImputer` de *Scikit-learn*).  

````python
# Forward fill
df["age_ffill"] = df["age"].ffill()

# Backward fill
df["age_bfill"] = df["age"].bfill()

df
````

## 6. Filtrado e indexación

Una de las características más potentes de Pandas es la capacidad de **seleccionar** y **filtrar** datos de forma flexible.

Trabajar con datos significa, casi siempre, **seleccionar subconjuntos** de un DataFrame:  
- Filtrar filas que cumplen ciertas condiciones.  
- Seleccionar columnas específicas.  
- Acceder a posiciones concretas.  
- Modificar valores en función de un criterio.  

En Pandas existen varias formas de hacerlo: **operadores lógicos, `.loc[]`, `.iloc[]`, y acceso directo a columnas**.

### <u>Filtrado con condiciones lógicas</u>

El filtrado más común consiste en aplicar condiciones sobre columnas:

````python
# Pasajeros mayores de 30 años
df[df["age"] > 30]

# Mujeres en primera clase
# df[(df["sex"] == "female") & (df["pclass"] == 1)]

# Pasajeros con tarifa > 50 o menores de 18
# df[(df["fare"] > 50) | (df["age"] < 18)]

# Pasajeros que NO embarcaron en Southampton
# df[~(df["embarked"] == "S")]
````

> **Operadores disponibles:**
>- `&` → AND lógico (ambas condiciones se cumplen).
>- `|` → OR lógico (se cumple al menos una condición).
>- `~` → NOT lógico (invierte la condición).

⚠️ Importante: hay que usar paréntesis en cada condición, si no Pandas lanza error de precedencia.

### <u>Selección de columnas</u>

Existen varias formas de seleccionar columnas:

````python
# Una columna (devuelve Series)
df["age"]

# Varias columnas (devuelve DataFrame)
# df[["age", "sex", "fare"]]
````

> La selección de columnas puede combinarse con condiciones de filas:

````python
# Mujeres menores de 20, solo columnas edad y tarifa
df[(df["sex"] == "female") & (df["age"] < 20)][["age", "fare"]]
````

### <u>Selección por etiquetas `.loc[]`</u>

`.loc[]` permite seleccionar **filas y columnas por nombre/etiqueta**

Sintaxis:

```python
df.loc[filas, columnas]

````python
# Fila con índice 0
df.loc[0]

# Filas del 0 al 4 y columnas "sex" y "age"
# df.loc[0:4, ["sex", "age"]]

# Pasajeros mujeres, seleccionando edad y tarifa
# df.loc[df["sex"] == "female", ["age", "fare"]]
````

> Características:
>- El rango de índices es inclusivo (0:4 incluye la fila 4).
>- Se puede usar con condiciones, listas de índices o etiquetas.

### <u>Selección por posicion `.iloc[]`</u>

`.iloc[]` selecciona por **posición numérica** (similar a arrays NumPy).

Sintaxis:

```python
df.iloc[filas, columnas]

````python
# Primera fila
df.iloc[0]

# Primeras 3 filas y primeras 2 columnas
# df.iloc[0:3, 0:2]

# Elemento en fila 5, columna 3
# df.iloc[5, 3]
````

> Características:
>- El rango es exclusivo en el final, como en Python (0:3 incluye filas 0,1,2).
>- Muy útil cuando no conoces los nombres de las columnas, solo su posición.

### <u>Diferencias clave entre `.loc[]` y `.iloc[]`</u>

| Método    | Selección                    | Ejemplo              | Inclusividad         |
| --------- | ---------------------------- | -------------------- | -------------------- |
| `.loc[]`  | Por **etiquetas** (nombres)  | `df.loc[0:2, "age"]` | Incluye índice final |
| `.iloc[]` | Por **posiciones** (números) | `df.iloc[0:2, 3]`    | Excluye índice final |


> Regla práctica:
>- Usa `.loc[]` si trabajas con nombres de filas o columnas.
>- Usa `.iloc[]` si trabajas con posiciones numéricas.

## 8. Combinación y Merge de DataFrames

En proyectos de IA y ciencia de datos es muy común tener la información repartida en varios ficheros o tablas.  
Pandas ofrece herramientas para **combinar y unir DataFrames** de forma muy parecida a las operaciones de **joins en SQL** o a concatenar hojas de cálculo en Excel.

Las dos funciones principales son:

- **`merge()`** → une DataFrames según columnas o índices comunes (como los *joins* en SQL).  
- **`concat()`** → pega DataFrames uno encima de otro (filas) o uno al lado del otro (columnas).  

### <u>`merge()`: unir DataFrames como en SQL</u>

El método `merge()` permite combinar dos DataFrames usando una clave común.

#### Sintaxis básica
```python
pd.merge(df1, df2, on="columna_clave")
```

**Tipos de joins disponibles**:
- *inner* (por defecto): se queda solo con los registros comunes en ambas tablas.
- *left*: conserva todas las filas del DataFrame izquierdo.
- *right*: conserva todas las filas del DataFrame derecho.
- *outer*: conserva todas las filas de ambos, rellenando con NaN lo que falte.

````python
import pandas as pd

# DataFrame con pasajeros y edades
df1 = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "nombre": ["Ana", "Luis", "Marta", "Carlos"],
    "edad": [23, 34, 29, 40]
})

# DataFrame con pasajeros y ciudades
df2 = pd.DataFrame({
    "id": [2, 3, 4, 5],
    "ciudad": ["Madrid", "Barcelona", "Valencia", "Sevilla"]
})

display(df1)
display(df2)
````

````python
# Inner join
pd.merge(df1, df2, on="id", how="inner")
````

````python
# Left join
pd.merge(df1, df2, on="id", how="left")
````

````python
# Outer join
pd.merge(df1, df2, on="id", how="outer")
````

### <u>`concat()`: concatenar DataFrames</u>

La función `concat()` se utiliza para **apilar DataFrames**:

- Por filas (uno debajo del otro).  
- Por columnas (uno al lado del otro).  

#### Sintaxis básica
```python
pd.concat([df1, df2], axis=0)  # por filas
pd.concat([df1, df2], axis=1)  # por columnas
```

````python
df_a = pd.DataFrame({"A": [1, 2], "B": [3, 4]})
df_b = pd.DataFrame({"A": [5, 6], "B": [7, 8]})

display(df_a)
display(df_b)
````

````python
# Concatenación por filas
pd.concat([df_a, df_b], axis=0)
````

````python
# Concatenación por columnas
df_c = pd.DataFrame({"C": ["x", "y"]})
pd.concat([df_a, df_c], axis=1)
````

## 9. Optimización y Buenas Prácticas en Pandas

Cuando trabajamos con datasets pequeños (ej. Titanic) casi no importa la eficiencia.  
Pero en proyectos reales de IA podemos tener **millones de filas** y **decenas de columnas**, lo que exige escribir código en Pandas que sea:

- **Más eficiente en memoria.**  
- **Más rápido en ejecución.**  
- **Escalable a datos grandes.**

En esta sección veremos algunas técnicas clave.

### <u>Reducción de memoria</u>

Cada columna en Pandas tiene un tipo de datos (`dtype`).  
Si usamos un tipo más grande de lo necesario, gastamos memoria innecesaria.

#### Estrategias:
- Convertir columnas con pocos valores únicos a **`category`**.  
- Usar enteros más pequeños (`int8`, `int16` en lugar de `int64`).  
- Usar decimales más pequeños (`float32` en lugar de `float64`).

````python
import pandas as pd

df = pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"))

print("Uso de memoria original:")
print(df.memory_usage(deep=True))

print("-"*150)

# Optimización
df_opt = df.copy()
df_opt["pclass"] = df_opt["pclass"].astype("int8")
df_opt["sibsp"] = df_opt["sibsp"].astype("int8")
df_opt["parch"] = df_opt["parch"].astype("int8")
df_opt["fare"] = df_opt["fare"].astype("float32")
df_opt["sex"] = df_opt["sex"].astype("category")
df_opt["embarked"] = df_opt["embarked"].astype("category")

print("\nUso de memoria optimizado:")
print(df_opt.memory_usage(deep=True))
````

### <u>Lectura parcial de datos</u>

No siempre es necesario cargar todo el dataset en memoria.

#### Opciones:
- **`nrows`** → lee solo un número limitado de filas.  
- **`chunksize`** → divide la lectura en bloques más pequeños (*streaming*).  

````python
# Leer solo las primeras 5 filas
df_small = pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"), nrows=5)
print(df_small)

print("-"*150)

# Leer en chunks de 5 filas
for chunk in pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"), chunksize=5):
    print("Chunk leído:")
    print(chunk)
````

### <u>Vectorización vs. `apply`</u>

Pandas está diseñado para operar sobre **columnas enteras** (vectorización).  
Esto es mucho más rápido que aplicar funciones fila a fila.

#### Ejemplo:

````python
# Vectorizado (rápido)
df["fare_doble"] = df["fare"] * 2

# Con apply (más lento)
df["fare_doble_apply"] = df["fare"].apply(lambda x: x * 2)

df[["fare", "fare_doble", "fare_doble_apply"]].head()
````

<u>Regla general</u>: **usa operaciones vectorizadas siempre que sea posible**.  
`apply` solo debe usarse cuando la lógica no pueda expresarse de forma vectorizada.

### <u>Medición de tiempos con `%timeit`</u>

En Jupyter Notebook podemos usar **`%timeit`** para medir tiempos de ejecución de una celda o línea de código.
    
Esto nos ayuda a comparar enfoques y elegir el más eficiente.

Normalmente el vectorizado es entre **10 y 100 veces más rápido** que usar `apply`.

````python
# Ejemplo: comparar vectorizado vs apply
import numpy as np

# Generamos una columna grande para ver diferencias
df_big = pd.DataFrame({"valores": np.random.rand(1_000_000)})

# Vectorizado
%timeit df_big["valores"] * 2

# Con apply
%timeit df_big["valores"].apply(lambda x: x * 2)
````

## 10. Preparación de Datos para IA

Antes de entrenar un modelo de Machine Learning, debemos transformar los datos en un formato **numérico, limpio y normalizado**.  
Los algoritmos no entienden directamente cadenas de texto ni categorías; por eso necesitamos aplicar **preprocesamiento**.

En esta sección veremos:
- Selección de columnas (features relevantes).  
- Codificación de variables categóricas (one-hot encoding, label encoding).  
- Normalización y escalado numérico.  
- Obtención de un DataFrame listo para modelado.

### <u>Selección de columnas y features relevantes</u>

Un **feature** es una variable que usaremos como entrada en un modelo.  
No todas las columnas de un DataFrame son útiles para entrenar:

- Algunas son identificadores (`id`, `nombre`) → no aportan información.  
- Otras son redundantes o derivadas.  
- Debemos decidir qué columnas usar como **X** (features) y cuál como **y** (target).  

Ejemplo con Titanic reducido:

````python
df = pd.read_csv(os.path.join(os.getcwd(), "titanic_demo", "titanic_demo.csv"))

# Target = columna a predecir
y = df["survived"]

# Features seleccionadas
X = df[["pclass", "sex", "age", "fare", "embarked"]]
X.head()
````

### <u>One-hot encoding con `pd.get_dummies()`</u>

Los algoritmos no entienden variables categóricas (`sex`, `embarked`).  
Una opción es usar **One-Hot Encoding**, que crea una columna binaria (0/1) por cada categoría.

En Pandas podemos hacerlo fácilmente con `pd.get_dummies()`.

````python
X_encoded = pd.get_dummies(X, columns=["sex", "embarked"])
X_encoded
````

### <u>LabelEncoder y MinMaxScaler</u>

Para tareas de preprocesamiento más avanzadas solemos usar **scikit-learn**.  

- **LabelEncoder** → convierte categorías en enteros (`male=0`, `female=1`).  
- **MinMaxScaler** → normaliza valores numéricos a un rango, normalmente `[0,1]`.  

Esto es útil para algoritmos sensibles a la escala de las variables (ej. redes neuronales).

````python
from sklearn.preprocessing import LabelEncoder, MinMaxScaler

# Copiamos el DataFrame
X_ml = X.copy()

# LabelEncoder en 'sex' y 'embarked'
le_sex = LabelEncoder()
X_ml["sex"] = le_sex.fit_transform(X_ml["sex"])

le_emb = LabelEncoder()
X_ml["embarked"] = le_emb.fit_transform(X_ml["embarked"].astype(str))

# MinMaxScaler en columnas numéricas
scaler = MinMaxScaler()
X_ml[["age", "fare"]] = scaler.fit_transform(X_ml[["age", "fare"]])

X_ml.head()
````

### <u>DataFrame listo para modelado</u>

Después de estas transformaciones, tenemos un DataFrame **totalmente numérico**, sin valores categóricos ni escalas inconsistentes.

Este `X_ml` se puede usar directamente en cualquier modelo de Machine Learning:

- Árboles de decisión, Random Forest, Gradient Boosting.  
- Redes neuronales en TensorFlow/PyTorch.  
- Regresiones y clasificadores en Scikit-learn.
