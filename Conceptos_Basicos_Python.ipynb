### range<a href="#range" class="anchor-link">¶</a>

Función para generar secuencias. \</p\>range (b): Secuencia de valores
entre 0 y b (0 incluido, b excluido). \</p\>range (a,b): Secuencia de
valores entre a y b (a incluido, b excluido). \</p\>range (a,b,n):
Secuencia de valores entre a y b (a incluido, b excluido) con n saltos

```python
"""
instrucciones 1
for variable in secuencia:
    instrucciones 2
instrucciones 3
"""
for i in range(10):
    print(i)
```

```python
"""
instrucciones 1
while condición:
    instrucciones 2
instrucciones 3
"""
i=0
while i<10:
    print(i)
    i=i+1
```
