## 2. Introducción a la visualización de datos

La **visualización de datos** es una de las herramientas más poderosas en el flujo de trabajo de **ciencia de datos e inteligencia artificial**.  
No se trata solo de “hacer gráficos bonitos”, sino de **entender, depurar y comunicar** la información de manera eficaz.

---

### <u>Rol de la visualización en ciencia de datos e IA</u>

- **Exploración de Datos (EDA)**:  
  Antes de entrenar cualquier modelo, necesitamos **explorar el dataset**.  
  Los gráficos nos permiten:
  - Detectar valores atípicos (*outliers*).  
  - Entender distribuciones de variables.  
  - Identificar correlaciones entre atributos.  

- **Depuración y diagnóstico de modelos**:  
  Durante el entrenamiento, los gráficos ayudan a:
  - Visualizar la **curva de aprendizaje** (train vs validation).  
  - Analizar errores: ejemplos mal clasificados, distribuciones de residuos.  
  - Evaluar métricas a lo largo del tiempo.  

- **Comunicación de resultados**:  
  En ciencia de datos, no basta con tener un buen modelo.  
  Es fundamental **explicar los resultados a otras personas** (científicos, directivos, clientes).  
  Una buena visualización:
  - Resume información compleja en segundos.  
  - Facilita la toma de decisiones.  
  - Da transparencia y confianza en el modelo.  

---

### <u>Ejemplos reales de uso</u>

1. **Exploración de datos en salud**  
   - Un histograma de edades revela sesgos en la población estudiada.  
   - Un gráfico de dispersión muestra relación entre presión arterial y colesterol.

2. **Depuración de un modelo de IA**  
   - En visión por computador, mostrar las imágenes donde el modelo falla ayuda a entender limitaciones.  
   - En NLP, un gráfico de nubes de palabras puede mostrar qué términos confunden al clasificador.  

3. **Comunicación en empresas**  
   - Dashboards con gráficas interactivas permiten a los directivos monitorizar métricas clave.  
   - Visualizaciones temporales (series de tiempo) muestran tendencias y ayudan a planificar.

## 3. Plotly Basics

En Plotly trabajaremos con dos APIs:

- **`plotly.express` (px)** → sintaxis **rápida y sencilla** para gráficos comunes.
- **`plotly.graph_objects` (go)** → API **flexible y granular** para personalizar cada detalle.

Además veremos **personalización** (títulos, ejes, leyendas, colores) aplicable a ambas.

### <u>plotly.express (rápido y sencillo)</u>

`plotly.express` permite crear gráficos con una sola llamada:
- El mapeo de columnas se hace con argumentos (`x=`, `y=`, `color=`, `size=`, etc.)
- Devuelve una figura Plotly que puedes seguir personalizando con `.update_layout()` y `.update_traces()`.

````python
# Dataset de ejemplo
np.random.seed(7)
df = pd.DataFrame({
    "edad": np.random.randint(18, 70, 300),
    "ingresos": np.random.normal(2500, 600, 300).round(0),
    "region": np.random.choice(["Norte", "Sur", "Este", "Oeste"], 300, p=[.25,.25,.25,.25]),
    "genero": np.random.choice(["Mujer", "Hombre"], 300)
})

display(df)
````

````python
# Histograma rápido
fig = px.histogram(df, x="edad", nbins=25, color="genero",
                   title="Distribución de edades por género")
fig.show()

# Dispersión con color y tamaño
fig = px.scatter(df, x="edad", y="ingresos", color="region", trendline="ols",
                 title="Edad vs Ingresos por región", opacity=0.8)
fig.show()

# Boxplot por categoría
fig = px.box(df, x="region", y="ingresos", color="region",
             title="Ingresos por región (Boxplot)")
fig.show()
````

### <u>plotly.graph_objects (flexible y personalizable)</u>

`graph_objects` expone **clases de trazas** (p.ej. `go.Scatter`, `go.Bar`, `go.Box`) y la figura `go.Figure`.
- Controlas cada capa y cada atributo.
- Ideal para gráficos con varios sub-trazos y personalización fina.

````python
import plotly.graph_objects as go

x = np.linspace(0, 2*np.pi, 200)
y1 = np.sin(x)
y2 = np.cos(x)

fig = go.Figure()

# Traza 1 (línea)
fig.add_trace(go.Scatter(x=x, y=y1, mode="lines", name="sin(x)"))

# Traza 2 (línea punteada)
fig.add_trace(go.Scatter(x=x, y=y2, mode="lines", name="cos(x)", line=dict(dash="dash")))

# Personalización básica (ver sección 3 para más)
fig.update_layout(
    title="Seno y Coseno",
    xaxis_title="x (rad)",
    yaxis_title="valor",
    template="plotly_white"
)
fig.show()
````

### <u>Personalización: títulos, ejes, leyendas, colores</u>

Casi todo se puede ajustar vía:
- **`fig.update_layout(...)`** → título, fuentes, márgenes, leyenda, fondo, plantillas.
- **`fig.update_xaxes(...) / fig.update_yaxes(...)`** → títulos y formato de ejes.
- **`fig.update_traces(...)`** → estilo de líneas, marcadores, opacidad, texto.
- En `px.*`, puedes pasar **`color_discrete_sequence`** o usar **`template`**.

````python
# Partimos de un px.scatter y lo tuneamos
fig = px.scatter(df, x="edad", y="ingresos", color="region",
                 title="Edad vs Ingresos — Personalizado")

fig.show()

# # Layout
# fig.update_layout(
#     title=dict(text="Edad vs Ingresos — Personalizado", x=0.05, xanchor="left"),
#     legend_title_text="Región",
#     template="plotly_white",
#     margin=dict(l=60, r=40, t=60, b=50),
# )

# fig.show()

# # Ejes
# fig.update_xaxes(title_text="Edad (años)", showgrid=True, zeroline=False)
# fig.update_yaxes(title_text="Ingresos (€/mes)", showgrid=True, zeroline=False)

# fig.show()

# # Trazas (marcadores más grandes y semitransparentes)
# fig.update_traces(marker=dict(size=9, line=dict(width=0.5)), opacity=0.85)

# fig.show()
````

## 4. Tipos de gráficos esenciales

En ciencia de datos e inteligencia artificial, la visualización es una herramienta crítica 
para explorar patrones, comunicar hallazgos y validar hipótesis. Plotly ofrece un conjunto 
de **gráficos básicos pero poderosos** que cubren la mayoría de las necesidades en un análisis exploratorio de datos (EDA) o en un informe.

A continuación, estudiaremos los tipos de gráficos más esenciales.

````python
np.random.seed(42)

# Dataset sintético
df = pd.DataFrame({
    "edad": np.random.randint(18, 65, 150),
    "ingresos": np.random.normal(30000, 8000, 150),
    "region": np.random.choice(["Norte", "Sur", "Este", "Oeste"], 150),
    "fecha": pd.date_range("2023-01-01", periods=150, freq="D"),
})
````

### <u>Gráficos de dispersión (Scatter)</u>

- **Qué representan:** muestran la relación entre dos variables numéricas.  
- **Cuándo usarlos:** detectar correlaciones, tendencias o outliers.  
- **Ejes típicos:** variable independiente en el eje X, variable dependiente en el eje Y.  
- **Personalizaciones comunes:**
  - Tamaño de los puntos (`size`) para añadir una tercera variable.  
  - Color (`color`) para representar categorías o valores continuos.  
  - Tooltip con información adicional (`hover_data`). 

````python
fig = px.scatter(
    df, x="edad", y="ingresos",
    color="region",
    title="Relación entre edad e ingresos"
)
fig.show()
````

### <u>Series temporales (Line charts)</u>

- **Qué representan:** evolución de una variable a lo largo del tiempo.  
- **Cuándo usarlos:** seguimiento de métricas, ventas, temperaturas, valores de sensores, etc.  
- **Claves:**
  - El eje X siempre debe ser temporal (fecha/hora).  
  - Se pueden comparar múltiples series (una por categoría).  
  - Suavizado o *rolling averages* ayudan a ver tendencias.  

````python
fig = px.line(
    df, x="fecha", y="ingresos",
    title="Evolución temporal de ingresos"
)
fig.show()
````

### <u>Gráficos de barras (Bar charts)</u>

- **Qué representan:** comparan valores entre distintas categorías.  
- **Cuándo usarlos:** distribución de frecuencias, top-N, comparaciones por grupos.  
- **Variantes:**
  - Barras verticales u horizontales.  
  - Agrupadas (`barmode="group"`) o apiladas (`barmode="stack"`).  

````python
fig = px.bar(
    df, x="region", y="ingresos",
    title="Ingresos promedio por región",
    barmode="group"
)
fig.show()
````

### <u>Diagramas de caja (Boxplots)</u>

- **Qué representan:** resumen de la distribución de una variable numérica.  
- **Elementos clave:**
  - Mediana (línea central).  
  - Cuartiles (caja).  
  - Bigotes (rango típico).  
  - Outliers (puntos extremos).  
- **Cuándo usarlos:** comparar distribuciones entre grupos, detectar asimetrías y valores atípicos. 

````python
fig = px.box(
    df, x="region", y="ingresos", color="region",
    title="Distribución de ingresos por región"
)
fig.show()
````

### <u>Mapas de calor (Heatmaps)</u>

- **Qué representan:** intensidad de valores en una matriz (2 variables).  
- **Variantes en Plotly:**
  - `px.imshow`: para visualizar directamente una matriz o imagen.  
  - `px.density_heatmap`: para datos dispersos (frecuencias en bins de dos variables).  
- **Cuándo usarlos:** correlaciones, matrices de confusión, concentraciones de puntos. 

````python
# Matriz de correlación
corr = df[["edad", "ingresos"]].corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="Matriz de correlación"
)
fig.show()
````

## 5. Interactividad y análisis

Plotly permite crear gráficos **interactivos por defecto**, lo que hace posible explorar los datos
de forma intuitiva: hacer zoom, inspeccionar valores concretos, activar o desactivar categorías, etc.

Además, ofrece mecanismos adicionales como `hover_data`, sliders, subplots y faceting, que enriquecen el análisis.

### <u>`Hover data`</u>

Con `hover_data` podemos mostrar información adicional al pasar el ratón sobre un punto. Esto nos ayuda a identificar registros individuales dentro de un conjunto grande de datos.

````python
np.random.seed(42)
df = pd.DataFrame({
    "edad": np.random.randint(18, 65, 200),
    "ingresos": np.random.normal(30000, 8000, 200),
    "region": np.random.choice(["Norte", "Sur", "Este", "Oeste"], 200),
    "genero": np.random.choice(["Hombre", "Mujer"], 200),
    "fecha": pd.date_range("2023-01-01", periods=200, freq="D")
})

fig = px.scatter(
    df, x="edad", y="ingresos",
    color="region",
    hover_data=["genero", "fecha"],
    title="Scatter con información en hover"
)
fig.show()
````

### <u>Sliders</u>

Un slider en el eje X permite explorar series temporales dentro de un rango concreto. En Plotly se activa con `xaxis=dict(rangeslider=dict(visible=True))`.

````python
fig = px.line(
    df, x="fecha", y="ingresos",
    title="Evolución de ingresos con slider"
)

fig.update_layout(
    xaxis=dict(rangeslider=dict(visible=True))
)
fig.show()
````

### <u>Subplots</u>

Con `make_subplots` podemos disponer varios gráficos en la misma figura.  
Cada subplot mantiene su interactividad y se pueden combinar diferentes tipos de gráficos.

````python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(rows=1, cols=2, subplot_titles=("Histograma de edades", "Scatter edad vs ingresos"))

fig.add_trace(go.Histogram(x=df["edad"], name="Edades"), row=1, col=1)
fig.add_trace(go.Scatter(x=df["edad"], y=df["ingresos"], mode="markers", name="Scatter"), row=1, col=2)

fig.update_layout(title="Subplots interactivos", template="plotly_white")
fig.show()
````

### <u>Faceting</u>

El faceting permite dividir un gráfico en múltiples paneles, cada uno mostrando un subconjunto de datos filtrado por una categoría.  
Se consigue con los parámetros `facet_row` y/o `facet_col`.

````python
fig = px.box(
    df, x="region", y="ingresos", color="genero",
    facet_col="genero",
    title="Boxplots por región y género"
)
fig.show()
````

## 6. Dashboards

Hasta ahora hemos creado gráficos individuales, pero en un proyecto real de análisis o de IA 
normalmente necesitamos **combinar varias visualizaciones** en un mismo espacio para contar una historia completa.

Existen dos aproximaciones principales en el ecosistema Plotly:

### <u>`Dash`</u>

*Documentación oficial de Dash*: https://dash.plotly.com/

**Dash** es un framework en Python que permite crear **aplicaciones web interactivas** para desplegar gráficos, paneles y controles sin necesidad de programar en JavaScript o HTML.  
Es especialmente útil en proyectos de IA porque:

- Permite visualizar en tiempo real los resultados de un modelo (ej. predicciones de un clasificador).  
- Facilita la creación de dashboards para monitorizar métricas de entrenamiento o validación.  
- Ayuda a explorar conjuntos de datos y hacer **EDA interactivo**.  
- Hace posible entregar una interfaz sencilla para que usuarios no técnicos prueben modelos.

#### Características principales
- Cada gráfico de Plotly se convierte en un *componente web*.  
- Se pueden añadir controles como sliders, menús desplegables o checkboxes.  
- Es ideal para compartir resultados con usuarios no técnicos o construir paneles internos.

#### Arquitectura básica de Dash

Una aplicación Dash consta de dos partes principales:

1. **Layout (diseño de la interfaz)**  
   - Define qué se muestra en pantalla: gráficos, menús desplegables, sliders, botones, etc.  
   - Usa componentes de `dash.html` y `dash.dcc` (Dash Core Components).

2. **Callbacks (interactividad)**  
   - Definen la lógica de cómo los elementos reaccionan a las acciones del usuario.  
   - Se implementan como funciones Python decoradas con `@app.callback`.

#### Flujo de trabajo típico

1. Importar las librerías necesarias.  
2. Crear la aplicación con `app = dash.Dash(__name__)`.  
3. Definir el **layout** (qué se ve).  
4. (Opcional) Definir los **callbacks** (cómo reacciona).  
5. Ejecutar la app con `app.run()` y abrirla en el navegador.

#### Ejemplo básico

Supongamos que queremos analizar cómo se relacionan **PIB per cápita** y **esperanza de vida**, variables relevantes en modelos de predicción demográfica o sanitaria.

````python
pip install dash
````

````python
import dash
from dash import dcc, html
import plotly.express as px

# Dataset de ejemplo
df = px.data.gapminder().query("year == 2007")

# Gráfico interactivo
fig = px.scatter(
    df, x="gdpPercap", y="lifeExp",
    size="pop", color="continent",
    hover_name="country", log_x=True, size_max=60
)

# Inicializar la app
app = dash.Dash(__name__)

# Layout de la app
app.layout = html.Div([
    html.H1("Relación PIB vs Esperanza de Vida (Gapminder 2007)"),
    dcc.Graph(figure=fig)
])

# Ejecutar servidor (http://127.0.0.1:8050)
if __name__ == "__main__":
    app.run(debug=True)
````

Una vez ejecutada la celda, abrir el navegador en `http://127.0.0.1:8050/`

Este tipo de panel sería el primer paso para crear un dashboard de análisis exploratorio interactivo en un proyecto de IA.

````python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

# Supongamos que tenemos métricas de entrenamiento
epochs = [1, 2, 3, 4, 5]
loss = [0.9, 0.6, 0.4, 0.35, 0.3]
accuracy = [0.5, 0.65, 0.72, 0.78, 0.8]

fig = make_subplots(rows=1, cols=2, subplot_titles=("Loss", "Accuracy"))

fig.add_trace(
    go.Scatter(x=epochs, y=loss, mode="lines+markers", name="Loss"),
    row=1, col=1
)

fig.add_trace(
    go.Scatter(x=epochs, y=accuracy, mode="lines+markers", name="Accuracy"),
    row=1, col=2
)

fig.update_layout(title_text="Evolución del entrenamiento del modelo")


# Inicializar la app
app = dash.Dash(__name__)

# Layout de la app
app.layout = html.Div([
    html.H1("Evolución del entrenamiento del modelo"),
    dcc.Graph(figure=fig)
])

# Ejecutar servidor (http://127.0.0.1:8050)
if __name__ == "__main__":
    app.run(debug=True)
````

### <u>Callbacks en `Dash`</u>

Los **callbacks** permiten que los elementos de un dashboard reaccionen a la interacción del usuario.  
Se implementan con el decorador `@app.callback` y conectan:

- **Inputs**: los valores de los componentes que cambian (ej. un `Dropdown`, un `Slider`, un `Input`).  
- **Outputs**: las propiedades de los componentes que se actualizan (ej. los datos de un gráfico, el texto de un párrafo).  

La firma básica de un callback es:

```python
@app.callback(
    Output("id-del-componente-salida", "propiedad"),
    Input("id-del-componente-entrada", "propiedad")
)
def funcion_callback(valor_entrada):
    # Lógica para actualizar la salida
    return nuevo_valor

````python
import dash
from dash import dcc, html, Input, Output

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Ejemplo de Callback"),
    dcc.Slider(0, 10, 1, value=5, id="mi-slider"),
    html.Div(id="output-text")
])

@app.callback(
    Output("output-text", "children"),
    Input("mi-slider", "value")
)
def actualizar_texto(valor):
    return f"Has seleccionado el valor: {valor}"

if __name__ == "__main__":
    app.run(debug=True)
````

<u>*NOTA*:</u> En Dash, todos los componentes (como html.Div, html.H1, dcc.Graph, etc.) tienen propiedades.  
Una de las propiedades más comunes es `children`, que representa el contenido interno de un componente.

```python
html.Div("Hola mundo")
```

es equivalente a

```python
html.Div(children="Hola mundo")
```

Aquí, `"Hola mundo"` es el valor de la propiedad `children` del `Div`.

En un callback, si usamos:

```python
Output("output-text", "children")
```

significa: *"El valor que devuelva la función callback se usará como contenido interno (`children`) del componente con id='output-text'".*

<u>Resumen:</u>
>- `children` = contenido dentro del componente.
>- Si no lo especificas, Dash lo asume por defecto.
>- En callbacks, actualizar `children` significa cambiar dinámicamente lo que se muestra dentro del componente.

## 7. Buenas prácticas y diseño efectivo en visualización de datos

En proyectos de **Inteligencia Artificial y Ciencia de Datos**, una visualización no solo debe ser **bonita**, sino sobre todo **efectiva**: debe transmitir de forma clara la información relevante para la toma de decisiones.  
Un mal diseño puede llevar a **interpretaciones erróneas de los resultados de un modelo** o a **comunicaciones poco persuasivas**.

### <u>Elección del gráfico</u>

La primera decisión clave es **qué tipo de gráfico utilizar** según los datos y el mensaje que queremos transmitir:

- **Comparaciones**  
  - Gráfico de barras o columnas.  
  - Ejemplo en IA: comparar el número de aciertos por clase en un clasificador de imágenes.  

- **Distribuciones**  
  - Histogramas, diagramas de caja (boxplots), violín plots.  
  - Ejemplo en IA: analizar la distribución de errores de predicción.  

- **Relaciones entre variables**  
  - Dispersión (scatter), burbujas, mapas de calor (heatmaps).  
  - Ejemplo en IA: comparar predicciones frente a valores reales en regresión.  

- **Evolución temporal**  
  - Gráficos de línea o área.  
  - Ejemplo en IA: evolución de la pérdida y precisión durante el entrenamiento de una red neuronal.  

**Regla práctica**: *elige siempre el gráfico más simple que transmita la idea de manera clara*.

### <u>Paleta de colores</u>

El uso de color debe ser **informativo, no decorativo**. Algunas buenas prácticas:

- Utilizar paletas consistentes y accesibles (ej. `px.colors.qualitative` de Plotly).
- Evitar combinaciones que dificulten la lectura para personas con **daltonismo** (usar paletas colorblind-friendly como `px.colors.qualitative.Safe`).
- Reservar los colores llamativos para destacar lo más importante.
- Mantener coherencia entre gráficos (ej. el mismo color siempre para la misma clase).

### <u>Uso correcto de escalas y etiquetas</u>

Otro punto clave es la correcta elección de escalas, títulos y etiquetas para evitar confusiones:

- **Escalas**  
  - Usar escalas lineales, logarítmicas o normalizadas según el caso.  
  - Ejemplo: un eje logarítmico es apropiado cuando los valores del PIB per cápita varían mucho entre países.  

- **Etiquetas**  
  - Los ejes y leyendas deben tener nombres claros y unidades de medida.  
  - Evitar abreviaturas poco conocidas o nombres de variables crudos de dataset (`lifeExp` → “Esperanza de vida (años)”).  

- **Títulos**  
  - Explicar en una frase qué muestra el gráfico.  
  - Evitar títulos vagos como *“Resultados”*. Mejor: *“Evolución de la pérdida en entrenamiento y validación”*.
